##Distribución Uniforme Discreta:

Parámetros: Supongamos que quieres simular una distribución uniforme discreta entre 1 y 10.
Simulación:

```python
import random
muestra_uniforme_discreta = [random.randint(1, 10) for _ in range(1000)]

# Inicializa una variable para almacenar la suma
suma = 0

# Recorre el vector y suma los valores
for valor in muestra_uniforme_discreta:
    suma += valor


#media y varianza a través de las formulas:
Media = suma/1000

#Valor de la opercaión en el parentesis de la varianza--(Hayamos (variable - media) elevado al cuadrado)
resultado = [(valor - Media) ** 2 for valor in muestra_uniforme_discreta]
Varianza = sum(resultado) / 1000
```

```python

import numpy as np

# Supongamos que 'muestra_uniforme_discreta' es la muestra simulada de la distribución uniforme discreta
media_uniforme_discreta = np.mean(muestra_uniforme_discreta)
varianza_uniforme_discreta = np.var(muestra_uniforme_discreta, ddof=1)


# Imprimir los resultados
print("Media muestral (Uniforme Discreta):", media_uniforme_discreta, " -- Y la Media realizada a través de la fórmula:", Media)
print("Varianza muestral (Uniforme Discreta):", varianza_uniforme_discreta, " -- Y la Varianza realizada a través de la fórmula:", Varianza)
print("Como observamos, la diferencia entre los 2 métodos, sólo se percibe en milécimas en la varianza, pero de por sí son valores muy parecidas")
```

```python
def lanzar_dado():
  """
  Simula el lanzamiento de un dado.

  Devuelve un número entero entre 1 y 6, ambos inclusive.
  """
  return random.randint(1, 6)

def calcular_proporcion(lanzamientos, veces_5):
  """
  Calcula la proporción de veces que ha salido el número 5 en una serie de lanzamientos.

  Args:
    lanzamientos: Número total de lanzamientos.
    veces_5: Número de veces que ha salido el número 5.

  Devuelve:
    La proporción de veces que ha salido el número 5.
  """
  return veces_5 / lanzamientos
```

```python
def main():
  """
  Programa principal.
  """
  # Número de lanzamientos a realizar.
  lanzamientos = [2, 6, 10, 100, 1000]

  # Bucle para realizar los lanzamientos.
  for n in lanzamientos:
    # Número de veces que ha salido el número 5.
    veces_5 = 0

    # Bucle para realizar los lanzamientos.
    for i in range(n):
      # Lanzamos el dado.
      numero = lanzar_dado()

      # Si el número es 5, actualizamos la cuenta.
      if numero == 5:
        veces_5 += 1

    # Calculamos la proporción.
    proporcion = calcular_proporcion(n, veces_5)

    # Imprimimos los resultados.
    print(f"Número de lanzamientos: {n}")
    print(f"Proporción de veces que ha salido el número 5: {proporcion}")

if __name__ == "__main__":
  main()
```

```python
def generar_exponencial(lambda_, size):
    """Genera una muestra aleatoria de tamaño size de una distribución exponencial"""
    return -np.log(np.random.uniform(0, 1, size)) / lambda_

def main():
    n_muestras = 10000
    medias = []
    for i in range(n_muestras):
        data = generar_exponencial(1, 1000)
        media = np.mean(data)
        medias.append(media)

    plt.hist(medias, bins=100)
    plt.show()

if __name__ == "__main__":
    main()
```

Distribución uniforme

Si generamos 10000 muestras aleatorias de tamaño 1000 de una distribución uniforme, la media de cada muestra estará aproximadamente entre 0 y 1. Si graficamos la distribución de las medias obtenidas, veremos que se aproxima a una campana de Gauss, también conocida como distribución normal.

Distribución exponencial

Si generamos 10000 muestras aleatorias de tamaño 1000 de una distribución exponencial, la media de cada muestra estará aproximadamente igual al parámetro lambda. Si graficamos la distribución de las medias obtenidas, veremos que se aproxima a una distribución exponencial.

En ambos casos, la distribución de las medias generadas se aproxima a una distribución normal o exponencial, dependiendo de la distribución de la muestra original. Esto se debe a que la ley de los grandes números establece que la media de una muestra aleatoria de tamaño grande de cualquier distribución se aproxima a la media de la distribución teórica.

```python
# Función para simular datos
def simulate_data(n, lambda_):
    data = np.random.exponential(scale=1/lambda_, size=n)
    return data

# Función para realizar la estimación
def estimation(data, estimator):
    return estimator(data)

# Estimadores
def estimator1(data):
    return data[0]

def estimator2(data):
    return 1

def estimator3(data):
    return np.sum(data) / len(data)

def estimator4(data):
    return np.min(data)
```

```python
# Simulación y estimación
n_values = [10, 100, 1000]
lambdas = [2, 5, 10]
estimators = [estimator1, estimator2, estimator3, estimator4]
estimation_results = {}
```

```python
for lambda_ in lambdas:
    estimation_results[lambda_] = {}
    for n in n_values:
        estimation_results[lambda_][n] = []
        for estimator in estimators:
            estimation_results[lambda_][n].append(estimation(simulate_data(n, lambda_), estimator))
```

```python
# Gráfico de resultados
for lambda_ in lambdas:
    plt.figure(figsize=(15, 15))
    for i, n in enumerate(n_values):
        for j, estimator in enumerate(estimators):
            plt.subplot(3, 4, i*4 + j + 1)
            plt.title(f"λ={lambda_}, n={n}, {estimator.__name__}")
            plt.hist(estimation_results[lambda_][n][j], bins=20, density=True)
            plt.axvline(x=lambda_, color='r', linestyle='--')
            plt.xlabel("Estimación")
            plt.ylabel("Probabilidad")
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import random

def generate_exponential(scale, size):
    """Genera una muestra aleatoria de tamaño size de una distribución exponencial con parámetro scale"""
    u = np.random.uniform(0, 1, size)
    return -np.log(1 - u) * scale

def estimator_1(data):
    """Estimador 1: El primer elemento de la muestra"""
    return data[0]

def estimator_2(data):
    """Estimador 2: El promedio de los primeros dos elementos de la muestra"""
    return np.mean(data[:2])

def estimator_3(data):
    """Estimador 3: La suma de los inversos de los elementos de la muestra"""
    return np.sum(1/data)

def estimator_4(data):
    """Estimador 4: El mínimo elemento de la muestra"""
    return np.min(data)

lambd = 1

for n in [10, 100, 1000]:
    data = generate_exponential(scale=1/lambd, size=n)
    est_1 = estimator_1(data)
    est_2 = estimator_2(data)
    est_3 = estimator_3(data)
    est_4 = estimator_4(data)

    print(f"Para n = {n}:")
    print(f"Estimador 1: {est_1}")
    print(f"Estimador 2: {est_2}")
    print(f"Estimador 3: {est_3}")
    print(f"Estimador 4: {est_4}")

####El estimador 3, que toma la suma de los inversos de los elementos de la muestra
##como estimación del parámetro, se aproxima más al valor real del parámetro a medida que aumenta el tamaño de la muestra porque
##la suma de los inversos de los elementos de la muestra se acerca a la media de la distribución teórica, que es el parámetro.
```

```python
# Cargar los datos
df = pd.read_csv("SolicitudesDiarias.csv")

# Asegurarse de que las fechas estén en formato datetime
df['Fecha'] = pd.to_datetime(df['Fecha'])

# Definir las fechas de inicio de las campañas
fecha_primera_campana = pd.to_datetime("2022-02-01")
fecha_segunda_campana = pd.to_datetime("2022-06-01")

# Dividir los datos en tres grupos: antes de las campañas, después de la primera campaña y después de la segunda campaña
antes_campana = df[df['Fecha'] < fecha_primera_campana]
despues_primera_campana = df[(df['Fecha'] >= fecha_primera_campana) & (df['Fecha'] < fecha_segunda_campana)]
despues_segunda_campana = df[df['Fecha'] >= fecha_segunda_campana]


# Calcular el promedio diario antes de las campañas
promedio_antes = antes_campana['Solicitudes'].mean()

# Calcular el promedio diario después de la primera campaña
promedio_despues_primera = despues_primera_campana['Solicitudes'].mean()

# Calcular el promedio diario después de la segunda campaña
promedio_despues_segunda = despues_segunda_campana['Solicitudes'].mean()

'''
Realizar pruebas de hipótesis para comparar los promedios
'''

#Prueba para la Primera Campaña:
#Hipótesis Nula (H0): El promedio diario antes de la primera campaña es igual al promedio diario después de la primera campaña.
#Hipótesis Alternativa (H1): El promedio diario antes de la primera campaña es diferente del promedio diario después de la primera campaña.

t_stat_primera, p_valor_primera = stats.ttest_ind(antes_campana['Solicitudes'], despues_primera_campana['Solicitudes'])

#Prueba para la Segunda Campaña:

#Hipótesis Nula (H0): El promedio diario después de la primera campaña es igual al promedio diario después de la segunda campaña.
#Hipótesis Alternativa (H1): El promedio diario después de la primera campaña es diferente del promedio diario después de la segunda campaña

t_stat_segunda, p_valor_segunda = stats.ttest_ind(antes_campana['Solicitudes'], despues_segunda_campana['Solicitudes'])

# Calcular intervalos de confianza al 95%
confianza = 0.95
alpha = 1 - confianza

intervalo_antes = stats.t.interval(confianza, len(antes_campana) - 1, loc=np.mean(antes_campana['Solicitudes']), scale=stats.sem(antes_campana['Solicitudes']))
intervalo_primera = stats.t.interval(confianza, len(despues_primera_campana) - 1, loc=np.mean(despues_primera_campana['Solicitudes']), scale=stats.sem(despues_primera_campana['Solicitudes']))
intervalo_segunda = stats.t.interval(confianza, len(despues_segunda_campana) - 1, loc=np.mean(despues_segunda_campana['Solicitudes']), scale=stats.sem(despues_segunda_campana['Solicitudes']))

# Visualización de datos
plt.figure(figsize=(10, 6))
plt.plot(antes_campana['Fecha'], antes_campana['Solicitudes'], label="Antes de las campañas")
plt.plot(despues_primera_campana['Fecha'], despues_primera_campana['Solicitudes'], label="Después de la primera campaña")
plt.plot(despues_segunda_campana['Fecha'], despues_segunda_campana['Solicitudes'], label="Después de la segunda campaña")
plt.title("Solicitudes Diarias")
plt.xlabel("Fecha")
plt.ylabel("Cantidad")
plt.legend()
plt.show()

# Presentar resultados
print(f"Promedio diario antes de las campañas: {promedio_antes}")
print(f"Promedio diario después de la primera campaña: {promedio_despues_primera}")
print(f"Promedio diario después de la segunda campaña: {promedio_despues_segunda}")

print(f"Prueba de hipótesis para la primera campaña: Estadística t = {t_stat_primera}, p-valor = {p_valor_primera}")
print(f"Prueba de hipótesis para la segunda campaña: Estadística t = {t_stat_segunda}, p-valor = {p_valor_segunda}")

print(f"Intervalo de confianza al 95% para antes de las campañas: {intervalo_antes}")
print(f"Intervalo de confianza al 95% para después de la primera campaña: {intervalo_primera}")
print(f"Intervalo de confianza al 95% para después de la segunda campaña: {intervalo_segunda}")
```
